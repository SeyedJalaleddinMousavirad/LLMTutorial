```python
import torch
from datasets import load_dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM,
    GenerationConfig,
)
```

```python
huggingface_dataset_name = "knkarthick/dialogsum"

dataset = load_dataset(huggingface_dataset_name)
```

```python
example = dataset["test"][0]

print("DIALOGUE:")
print(example["dialogue"])

print("\nHUMAN SUMMARY:")
print(example["summary"])
```

```python
model_name = "google/flan-t5-base"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
model = AutoModelForSeq2SeqLM.from_pretrained(
    model_name,
    dtype=torch.float16
)
```

```python
model = model.to("cuda")
```

```python
print("Model device:", next(model.parameters()).device)
print("Model dtype:", next(model.parameters()).dtype)
```

```python
sentence = "What time is it, Tom?"

encoded = tokenizer(
    sentence,
    return_tensors="pt"
)

print(encoded)
```

```python
print(encoded["input_ids"])
```

```python
decoded = tokenizer.decode(
    encoded["input_ids"][0],
    skip_special_tokens=True
)

print(decoded)
```

```python
index = 0

dialogue = dataset["test"][index]["dialogue"]
human_summary = dataset["test"][index]["summary"]

print("DIALOGUE:")
print(dialogue)

print("\nHUMAN SUMMARY:")
print(human_summary)
```

```python
prompt = f"""
Summarize the following conversation:

{dialogue}

Summary:
"""

print(prompt)
```

```python
with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=100
    )
```

```python
model_summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(model_summary)
```

```python
print("HUMAN SUMMARY:")
print(human_summary)

print("\nMODEL SUMMARY:")
print(model_summary)
```

```python
example_indices = [0, 10, 20]

for i in example_indices:
    dialogue = dataset["test"][i]["dialogue"]
    human_summary = dataset["test"][i]["summary"]

    prompt = f"""
Summarize the following conversation:

{dialogue}

Summary:
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to("cuda")

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=100
        )

    model_summary = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    print("=" * 80)
    print(f"EXAMPLE {i}")

    print("\nDIALOGUE:")
    print(dialogue)

    print("\nHUMAN SUMMARY:")
    print(human_summary)

    print("\nMODEL SUMMARY:")
    print(model_summary)
```

```python
#a different prompt
dialogue = dataset["test"][0]["dialogue"]
human_summary = dataset["test"][0]["summary"]

prompt = f"""
Dialogue:

{dialogue}

What was going on?
"""

inputs = tokenizer(
    prompt,
    return_tensors="pt"
).to("cuda")

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=100
    )

model_summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("HUMAN SUMMARY:")
print(human_summary)

print("\nMODEL SUMMARY:")
print(model_summary)
```

```python
#One shot prompting
example_index = 10
test_index = 20

example_dialogue = dataset["test"][example_index]["dialogue"]
example_summary = dataset["test"][example_index]["summary"]

test_dialogue = dataset["test"][test_index]["dialogue"]
test_summary = dataset["test"][test_index]["summary"]


one_shot_prompt = f"""
Dialogue:

{example_dialogue}

Summary:
{example_summary}

Dialogue:

{test_dialogue}

Summary:
"""

print(one_shot_prompt)

inputs = tokenizer(
    one_shot_prompt,
    return_tensors="pt"
).to("cuda")

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=100
    )

one_shot_summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("HUMAN SUMMARY:")
print(test_summary)

print("\nONE-SHOT MODEL SUMMARY:")
print(one_shot_summary)
```

```python
#Few-shot prompting

example_indices = [10, 20]
test_index = 30

few_shot_prompt = ""

for i in example_indices:
    dialogue = dataset["test"][i]["dialogue"]
    summary = dataset["test"][i]["summary"]

    few_shot_prompt += f"""
Dialogue:

{dialogue}

Summary:
{summary}

"""

test_dialogue = dataset["test"][test_index]["dialogue"]

few_shot_prompt += f"""
Dialogue:

{test_dialogue}

Summary:
"""
print(few_shot_prompt)

inputs = tokenizer(
    few_shot_prompt,
    return_tensors="pt"
).to("cuda")

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=100
    )

few_shot_summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("HUMAN SUMMARY:")
print(dataset["test"][test_index]["summary"])

print("\nFEW-SHOT MODEL SUMMARY:")
print(few_shot_summary)
```
